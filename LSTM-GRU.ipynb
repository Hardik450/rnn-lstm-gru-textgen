```python
import tensorflow as tf

vocab_size = 20000
max_length = 1000

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.imdb.load_data(num_words= vocab_size)

x_train = tf.keras.preprocessing.sequence.pad_sequences(x_train, maxlen = max_length)
x_test = tf.keras.preprocessing.sequence.pad_sequences(x_test, maxlen = max_length)
```

```python
def lstm_model():
  model = tf.keras.models.Sequential()
  model.add(tf.keras.layers.Embedding(input_dim=vocab_size, output_dim=128, input_length = max_length))
  model.add(tf.keras.layers.LSTM(64))
  model.add(tf.keras.layers.Dense(1, activation='sigmoid'))
  return model
```

```python
lstm_model = lstm_model()
lstm_model.compile(optimizer= 'adam', loss='binary_crossentropy', metrics = ['accuracy'])
lstm_model.fit(x_train, y_train, epochs = 10, batch_size = 64, validation_split = 0.2)
```

```python
def gru_model():
  model = tf.keras.models.Sequential()
  model.add(tf.keras.layers.Embedding(input_dim = vocab_size, output_dim = 128))
  model.add(tf.keras.layers.GRU(64))
  model.add(tf.keras.layers.Dense(1, activation = 'sigmoid'))
  return model
```

```python
gru_model = gru_model()
gru_model.compile(optimizer = 'adam', loss='binary_crossentropy', metrics = ['accuracy'])
gru_model.fit(x_train, y_train, epochs = 10, batch_size = 64, validation_split = 0.2)
```

```python
lstm_loss, lstm_accuracy = lstm_model.evaluate(x_test, y_test)
print(f"LSTM Model\nAccuracy: {lstm_accuracy}, Loss: {lstm_loss}")
```

```python
gru_loss, gru_accuracy = gru_model.evaluate(x_test, y_test)
print(f"GRU Model\nAccuracy: {gru_accuracy}, Loss: {gru_loss}")
```

```python
lstm_model.save("lstm_model.keras")
gru_model.save("gru_model.keras")
```
